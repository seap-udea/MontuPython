# Working with stars

```python
%matplotlib inline
import montu as mn
# Autoreload refreshes local copies of montu when you edit the source.
# Not available in Google Colab (Python 3.12+): IPython autoreload still imports removed module `imp`.
# Uncomment only in a local Jupyter notebook:
# %load_ext autoreload
# %autoreload 2
```

```python
estrellas_visibles = mn.Stars().get_stars(Vmag=[-2,4])
rionegro = mn.Observer(lon=-75,lat=6,height=2.5)
mtime = mn.Time('2024-05-01 19:00:00')
mtime
```

```python
estrellas_visibles.where_in_sky(at=mtime,observer=rionegro,inplace=True)

mn.PRINTDF(estrellas_visibles.data.head(2))
```

```python
mn.Util.where_in_sky(at=mtime,observer=rionegro)
```

```python
mtime
```

```python
az,el = mn.Util.where_in_sky(RA=6.770358,Dec=-16.751203,at=mtime,observer=rionegro)
mn.Util.dec2sex(az), mn.Util.dec2sex(el)
```
