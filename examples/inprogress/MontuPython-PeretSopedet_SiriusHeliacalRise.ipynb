# Calculando la salida helíaca de Sirio

El objetivo de este notebook es calcular la salida helíaca de sirio (Peret Sopedet)

Lo haremos primero para el año de la tercer Apokatastasis

```python
historical_dates = montu.load_historical_dates()
historical_dates

```

```python
historical_dates['139-07-20']
```

```python
Tebas = montu.Observer(lon=33,lat=24,height=75)
Tebas
```

```python
td = t = montu.Time('139-07-20 00:00:00', calendar='mixed')
td
```

Vamos a calcular el tiempo de inicio del crepúsculo para ese día:

```python
dusk_time, dawn_time = montu.Sun.when_is_twilight(t, observer=Tebas, sunbelow=-11)
Tebas.get_local_time(dusk_time), Tebas.get_local_time(dawn_time) 
```

```python
dusk_time
```

```python
sirius = montu.Stars(subset='bright',ProperName='Sirius')
sirius.where_in_sky(montu.Time(dusk_time,format='jd',scale='utc'),Tebas,inplace=True)
sirius.data.el
```
