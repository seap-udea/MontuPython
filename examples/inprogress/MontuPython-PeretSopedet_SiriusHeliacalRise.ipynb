# Salida helíaca de Sirio (Peret Sopedet)

Usamos `montu.HeliacalRise` del módulo `montu.phenomena` para predecir la primera mañana visible de Sirio alrededor de la tercera *apokatastasis* (139 d.C.) en Tebas. Cada modelo se prueba por separado.

Parámetros comunes del barrido: `scan_start` = 139-07-01, `scan_end` = 139-08-15, `k = 0.25` mag/masa de aire, magnitud límite al cénit 6.0.

```python
third_apok = montu.load_historical_dates()['139-07-20']
third_apok
```

```python
Tebas = montu.Observer(lon=33, lat=24, height=0.075)
sirius = montu.Stars(subset='bright', ProperName='Sirius')

scan_start = montu.Time('139-07-01 00:00:00', calendar='mixed')
scan_end = montu.Time('140-08-15 00:00:00', calendar='mixed')

print('Observador:', Tebas)
print('Barrido:', scan_start.readable.datemix, '→', scan_end.readable.datemix)
```

## Schaefer 1987 — depresión solar fija

Criterio operativo con Sol a −11° y masa de aire secante. Fuente: Schaefer (1987), *Archaeoastronomy* 11.

```python
calc_1987 = montu.HeliacalRise(
    model='schaefer1987',
    k=0.25,
    limiting_mag_zenith=6.0,
    sun_depression=-11.0,
)
rise_1987 = calc_1987.compute(sirius, Tebas, scan_start, scan_end)
calc_1987.print_rises(rise_1987, title='Schaefer 1987', body_label='Sirio')
```

## Schaefer 1985 — listado BASIC

Reconstrucción del programa publicado en *Sky & Telescope* 70, 261–263. Barrido matutino con brillo crepuscular y umbral fisiológico.

```python
calc_1985 = montu.HeliacalRise(
    model='schaefer1985',
    k=0.25,
    limiting_mag_zenith=6.0,
)
rise_1985 = calc_1985.compute(sirius, Tebas, scan_start, scan_end)
calc_1985.print_rises(rise_1985, title='Schaefer 1985 (BASIC)', body_label='Sirio')
```

## Belokrylov et al. 2011

Modelo empírico con extinción y corrección por cercanía al Sol. Fuente: *Bulgarian Astronomical Journal* 16, ecuaciones (4)–(8).

```python
calc_2011 = montu.HeliacalRise(
    model='belokrylov2011',
    k=0.25,
    reference_extinction=0.25,
)
rise_2011 = calc_2011.compute(sirius, Tebas, scan_start, scan_end)
calc_2011.print_rises(rise_2011, title='Belokrylov 2011', body_label='Sirio')
```

```python
# %pip install montu
"""Heliacal-rise calculation with MontuPython Desktop default values."""

import montu

# Default observer location in MontuPython Desktop: Giza, Egypt.
observer = montu.Observer(lon=31.1342, lat=29.9792, height=0.075)

# Default body and search interval: Sirius, 139 CE, mixed calendar, one year.
sirius = montu.Stars(subset="bright", ProperName="Sirius")
start = montu.Time("139-01-01", calendar="mixed")
end = start + 365 * montu.DAY

# Default Schaefer (1987) visibility model parameters.
calculator = montu.HeliacalRise(
    model="schaefer1987",
    k=0.25,
    limiting_mag_zenith=6.0,
    sun_depression=-11.0,
)

events = calculator.compute(sirius, observer, start, end)
calculator.print_rises(
    events,
    title="Heliacal rises of Sirius",
    body_label="Sirius",
)
```

```python
## Grid search — Schaefer 1987 y los tres Apokatastasis

Buscamos parámetros del modelo `schaefer1987` (`k`, `limiting_mag_zenith`, `sun_depression`) que produzcan la salida helíaca de Sirio en **I-Akhet-1** para las tres fechas históricas de *apokatastasis*, en el sitio indicado.

**Criterio de ajuste:** priorizar aciertos en caniucular `I-Akhet-1`, luego coincidencia exacta con la fecha histórica, y minimizar el desfase en días.
```

```python
import time
from itertools import product

import numpy as np
import pandas as pd

# Sitio indicado para el ajuste (coordenadas del módulo Desktop / Giza).
Thebas = montu.Observer(lon=31.1342, lat=29.9792, height=0.075)
sirius = montu.Stars(subset='bright', ProperName='Sirius')

historical = montu.load_historical_dates()
APOKATASTASIS_KEYS = [
    'bce 2782-07-20',  # primera apokatastasis
    'bce 1322-07-20',  # segunda apokatastasis
    '139-07-20',       # tercera apokatastasis
]
APOKATASTASIS = [
    (key, historical[key]) for key in APOKATASTASIS_KEYS
]

SEARCH_BEFORE_DAYS = 90
SEARCH_AFTER_DAYS = 45


def first_heliacal_in_window(calculator, expected_key):
    """Primer heliacal rise detectado alrededor de la fecha histórica."""
    entry = historical[expected_key]
    expected = montu.Time(expected_key, calendar='mixed')
    start = expected - SEARCH_BEFORE_DAYS * montu.DAY
    end = expected + SEARCH_AFTER_DAYS * montu.DAY
    result = calculator.compute(sirius, Thebas, start, end)
    if result.empty:
        return {
            'historical_key': expected_key,
            'label': entry['label'],
            'expected_egyptian': entry['egyptian_date'],
            'detected_mixed': None,
            'detected_can': None,
            'day_error': np.nan,
            'akhet1_hit': False,
            'date_hit': False,
        }

    event = result.iloc[0]
    moment = montu.Time(event.day_jed, format='jd', calendar='mixed', full=True)
    day_error = abs(event.day_jed - expected.jed)
    detected_can = moment.readable.datecan
    return {
        'historical_key': expected_key,
        'label': entry['label'],
        'expected_egyptian': entry['egyptian_date'],
        'detected_mixed': moment.readable.datemix,
        'detected_can': detected_can,
        'day_error': day_error,
        'akhet1_hit': entry['egyptian_date'] in detected_can,
        'date_hit': day_error < 0.5,
        'local_time': event.local_time,
        'body_altitude_deg': event.body_altitude_deg,
        'sun_altitude_deg': event.sun_altitude_deg,
    }


def score_parameters(k, limiting_mag_zenith, sun_depression):
    calculator = montu.HeliacalRise(
        model='schaefer1987',
        k=float(k),
        limiting_mag_zenith=float(limiting_mag_zenith),
        sun_depression=float(sun_depression),
    )
    rows = [first_heliacal_in_window(calculator, key) for key, _ in APOKATASTASIS]
    akhet1_hits = sum(row['akhet1_hit'] for row in rows)
    date_hits = sum(row['date_hit'] for row in rows)
    total_day_error = float(np.nansum([row['day_error'] for row in rows]))
    score = akhet1_hits * 1000 + date_hits * 100 - total_day_error
    return {
        'k': float(k),
        'limiting_mag_zenith': float(limiting_mag_zenith),
        'sun_depression': float(sun_depression),
        'score': score,
        'akhet1_hits': akhet1_hits,
        'date_hits': date_hits,
        'total_day_error': total_day_error,
        'details': rows,
    }


def run_grid(k_values, lim_values, sun_values, label):
    started = time.perf_counter()
    rows = []
    total = len(k_values) * len(lim_values) * len(sun_values)
    for index, (k, lim, sun) in enumerate(product(k_values, lim_values, sun_values), start=1):
        rows.append(score_parameters(k, lim, sun))
        if index % max(1, total // 10) == 0:
            print(f'{label}: {index}/{total} combinaciones evaluadas …')
    df = pd.DataFrame(rows).sort_values(
        ['akhet1_hits', 'date_hits', 'total_day_error', 'score'],
        ascending=[False, False, True, False],
    )
    elapsed = time.perf_counter() - started
    print(f'{label} completado en {elapsed:.1f} s ({total} combinaciones).')
    return df


# --- Barrido grueso ---
coarse_k = np.arange(0.10, 0.55, 0.05)
coarse_lim = np.arange(3.5, 8.0, 0.5)
coarse_sun = np.arange(-16, -7, 1)

coarse_df = run_grid(coarse_k, coarse_lim, coarse_sun, 'Barrido grueso')
display(coarse_df.head(15))

best_coarse = coarse_df.iloc[0]
print('\nMejor combinación del barrido grueso:')
print(best_coarse[['k', 'limiting_mag_zenith', 'sun_depression', 'akhet1_hits', 'date_hits', 'total_day_error']])

# --- Refinamiento local alrededor del mejor punto ---
refine_k = np.arange(best_coarse.k - 0.05, best_coarse.k + 0.051, 0.01)
refine_lim = np.arange(best_coarse.limiting_mag_zenith - 0.5, best_coarse.limiting_mag_zenith + 0.51, 0.1)
refine_sun = np.arange(best_coarse.sun_depression - 1.0, best_coarse.sun_depression + 1.01, 0.25)

refined_df = run_grid(refine_k, refine_lim, refine_sun, 'Refinamiento')
display(refined_df.head(15))

best = refined_df.iloc[0]
print('\nMejor conjunto de parámetros encontrado:')
print(best[['k', 'limiting_mag_zenith', 'sun_depression', 'akhet1_hits', 'date_hits', 'total_day_error']])

verification_rows = []
for detail in best['details']:
    verification_rows.append({
        'Apokatastasis': detail['label'],
        'Fecha histórica': detail['historical_key'],
        'Esperado (caniucular)': detail['expected_egyptian'],
        'Detectado (mixed)': detail['detected_mixed'],
        'Detectado (caniucular)': detail['detected_can'],
        'Δ días': detail['day_error'],
        'I-Akhet-1': detail['akhet1_hit'],
        'Fecha exacta': detail['date_hit'],
    })

verification_df = pd.DataFrame(verification_rows)
display(verification_df)

print(
    f"\nResumen: {int(best.akhet1_hits)}/3 apokatastasis con I-Akhet-1; "
    f"{int(best.date_hits)}/3 con fecha histórica exacta."
)
if int(best.akhet1_hits) < 3:
    print(
        'Ningún conjunto del grid probado alcanzó I-Akhet-1 en las tres fechas '
        'con este observador y Schaefer 1987. El resultado anterior es el mejor compromiso.'
    )
```
